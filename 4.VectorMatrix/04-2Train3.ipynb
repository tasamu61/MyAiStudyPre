## データを読み込む

```python
class TrainingData:
    def __init__(self, filename, rows):
        self.filename = filename
        self.class_id = int(filename[0])   # 0="|", 1="-", 2="/"
        self.bits = [1.0 if ch == "■" else 0.0 for row in rows for ch in row]


training_data = [TrainingData(name, rows) for name, rows in RAW_DATA]

# 15件 × 256入力
X = [td.bits for td in training_data]
print("X件数 =", len(X), " 1件あたりの入力数 =", len(X[0]))

# 15件 × 3クラス。正解クラスの位置だけ1.0にする(one-hot)
targets = [[0.0, 0.0, 0.0] for _ in training_data]
for i, td in enumerate(training_data):
    targets[i][td.class_id] = 1.0

print("targets件数 =", len(targets), " 1件あたりのクラス数 =", len(targets[0]))
```

## モデル（3クラス分の重み・バイアス）

```python
# weights: 3クラス × 256入力、bias: 3クラス分
weights = [[0.0] * 256 for _ in range(3)]
bias = [0.0, 0.0, 0.0]

print("weights: 3 x 256, bias: 3")
```

## 学習ループ

numpy版の`X @ weights.T + bias`にあたる部分を、二重・三重のfor文で書く。

```python
rate = 0.04 / 4
epochs = 600

for epoch in range(epochs):

    # ①Prediction　15件 × 3クラス
    prediction = [[0.0, 0.0, 0.0] for _ in range(len(X))]
    for i in range(len(X)):
        for c in range(3):
            z = bias[c]
            for j in range(256):
                z += X[i][j] * weights[c][j]
            prediction[i][c] = z

    # ②diff, Loss
    diff = [[prediction[i][c] - targets[i][c] for c in range(3)] for i in range(len(X))]
    loss = sum(diff[i][c] ** 2 for i in range(len(X)) for c in range(3))

    # ③gradient（predictionに対する勾配）
    gradient = [[2 * diff[i][c] for c in range(3)] for i in range(len(X))]

    # ④weightのgradient（numpy版の gradient.T @ X にあたる部分）
    weight_gradient = [[0.0] * 256 for _ in range(3)]
    for c in range(3):
        for j in range(256):
            weight_gradient[c][j] = sum(gradient[i][c] * X[i][j] for i in range(len(X)))

    # ⑤biasのgradientと更新
    bias_gradient = [sum(gradient[i][c] for i in range(len(X))) for c in range(3)]

    for c in range(3):
        for j in range(256):
            weights[c][j] -= rate * weight_gradient[c][j]
        bias[c] -= rate * bias_gradient[c]

    if epoch % 50 == 0:
        print("epoch =", epoch, "loss =", loss)
```

## 学習結果

```python
print("----- result -----")

for i, td in enumerate(training_data):
    z = [0.0, 0.0, 0.0]
    for c in range(3):
        s = bias[c]
        for j in range(256):
            s += X[i][j] * weights[c][j]
        z[c] = s

    answer = 0
    for c in range(1, 3):
        if z[c] > z[answer]:
            answer = c

    print(td.filename, "target =", td.class_id,
          "prediction =", [round(v, 3) for v in z], "answer =", answer)
```
